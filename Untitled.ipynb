```python
import numpy as np
from sklearn.linear_model import TheilSenRegressor
from sklearn.utils import resample
from sklearn.datasets import make_regression

def fit_theilsen(X, y, random_state=0):
    m = TheilSenRegressor(random_state=random_state, fit_intercept=True)
    m.fit(X, y)
    return m

def bootstrap_coefs(X, y, n_boot=2000, random_state=0):
    rng = np.random.default_rng(random_state)
    n = len(y)
    coefs = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, size=n)
        m = fit_theilsen(X[idx], y[idx])
        coefs.append(m.coef_.copy())
    return np.array(coefs)  # shape: (n_boot, n_features)

def bootstrap_cis(boot_coefs, alpha=0.05, axis=0):
    boot_coefs = np.asarray(boot_coefs)
    if not np.isfinite(boot_coefs).all():
        raise ValueError("boot_coefs contains NaN or Inf")

    # If boot_coefs was built as (n_features, n_boot), fix axis
    if boot_coefs.ndim == 2 and boot_coefs.shape[0] < boot_coefs.shape[1]:
        axis = 1  # treat rows as features, cols as bootstrap draws

    lo_q = 100 * (alpha / 2)
    hi_q = 100 * (1 - alpha / 2)

    lo = np.percentile(boot_coefs, lo_q, axis=axis)
    hi = np.percentile(boot_coefs, hi_q, axis=axis)

    # Guard against accidental reversal
    lo, hi = np.minimum(lo, hi), np.maximum(lo, hi)
    return lo, hi

def bootstrap_pvalues(beta_hat, boot_coefs):
    # two sided: fraction with |beta_b| >= |beta_hat|
    return np.mean(np.abs(boot_coefs) >= np.abs(beta_hat), axis=0)

# --- Example usage on synthetic data with outliers ---
# Create a 2-feature problem
X, y, coef = make_regression(n_samples=300, n_features=2, noise=10, coef=True, random_state=7)
# add some outliers
rng = np.random.default_rng(42)
out_idx = rng.choice(len(y), size=15, replace=False)
y[out_idx] += rng.normal(0, 120, size=len(out_idx))

# --- Example usage ---
# X: (n_samples, n_features), y: (n_samples,)
# model fit
model = fit_theilsen(X, y)
beta_hat = model.coef_

# bootstrap
boot = bootstrap_coefs(X, y, n_boot=3000, random_state=1)
ci_lo, ci_hi = bootstrap_cis(boot, alpha=0.05)
p_boot = bootstrap_pvalues(beta_hat, boot)

print("coef:", beta_hat)
print("95% CI low:", ci_lo)
print("95% CI high:", ci_hi)
print("bootstrap p-values:", p_boot)
```
